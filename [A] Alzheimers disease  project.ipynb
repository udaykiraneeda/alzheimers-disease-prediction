```python
#=============================================
# Import Libraries
#=================================================
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# ============================================================
#  Load Dataset
# ============================================================
df = pd.read_csv("alzheimers_disease_data.csv")
df
```

```python
print("First 5 Rows")
print(df.head())
```

```python
print("\nDataset Shape")
print(df.shape)
```

```python
print("\nColumn Names")
print(df.columns.tolist())
```

```python
# ============================================================
# Drop Non-Informative Columns
# ============================================================

df.drop(columns=["PatientID", "DoctorInCharge"], inplace=True, errors="ignore")

print("\nShape After Dropping ID/Constant Columns")
print(df.shape)
```

```python
# ============================================================
#  Basic Information
# ============================================================

print("\nDataset Information")
print(df.info())
```

```python
print("\nData Types")
print(df.dtypes)
```

```python
# ============================================================
# Missing Values
# ============================================================

print("\nMissing Values")
print(df.isnull().sum())
```

```python
# ============================================================
# Duplicate Records
# ============================================================

print("\nDuplicate Rows =", df.duplicated().sum())

df.drop_duplicates(inplace=True)
```

```python
print("Shape After Removing Duplicates")
print(df.shape)
```

```python
# ============================================================
#  Statistical Summary
# ============================================================

print("\nStatistical Summary")
print(df.describe())
```

```python
# ============================================================
# Outlier Detection (Boxplots)
# ============================================================

continuous_cols = [
    "Age", "BMI", "AlcoholConsumption", "PhysicalActivity", "DietQuality",
    "SleepQuality", "SystolicBP", "DiastolicBP", "CholesterolTotal",
    "CholesterolLDL", "CholesterolHDL", "CholesterolTriglycerides",
    "MMSE", "FunctionalAssessment", "ADL"
]
```

```python
for col in continuous_cols:
    plt.figure(figsize=(6, 4))
    sns.boxplot(x=df[col])
    plt.title(f"Boxplot of {col}")
    plt.show()
```

```python
#============================================================
# Histograms
# ============================================================

df[continuous_cols].hist(figsize=(16, 12), bins=20)
plt.tight_layout()
plt.show()
```

```python
# ============================================================
# Distribution Plot
# ============================================================

for col in continuous_cols:
    plt.figure(figsize=(6, 4))
    sns.histplot(df[col], kde=True)
    plt.title(f"Distribution of {col}")
    plt.show()
```

```python
# ============================================================
#  Correlation Matrix
# ============================================================

corr = df.corr(numeric_only=True)
print(corr)
```

```python
# ============================================================
# Step 12: Heatmap
# ============================================================

plt.figure(figsize=(18, 14))
sns.heatmap(corr, annot=True, cmap="coolwarm", fmt=".2f", annot_kws={"size": 6})
plt.title("Correlation Heatmap")
plt.show()
```

```python
# ============================================================
# Step 13: Pair Plot
# ============================================================

pairplot_cols = ["Age", "BMI", "MMSE", "FunctionalAssessment", "ADL", "Diagnosis"]
sns.pairplot(df[pairplot_cols], hue="Diagnosis")
plt.show()
```

```python
# ============================================================
# Violin Plots
# ============================================================

for col in continuous_cols:
    plt.figure(figsize=(6, 4))
    sns.violinplot(y=df[col])
    plt.title(f"Violin Plot of {col}")
    plt.show()
```

```python
# ============================================================
# Scatter Plots
# ============================================================

sns.scatterplot(data=df, x="MMSE", y="FunctionalAssessment", hue="Diagnosis")
plt.title("MMSE vs Functional Assessment")
plt.show()

sns.scatterplot(data=df, x="Age", y="ADL", hue="Diagnosis")
plt.title("Age vs ADL")
plt.show()
```

```python
# ============================================================
#  Count Plot (Target Variable)
# ============================================================

plt.figure(figsize=(6, 4))
sns.countplot(x="Diagnosis", data=df)
plt.title("Diagnosis Class Distribution")
plt.show()
```

```python
# ============================================================
#  Class Percentage
# ============================================================

print(df["Diagnosis"].value_counts())
print(df["Diagnosis"].value_counts(normalize=True) * 100)
```

```python
# ============================================================
#  Skewness
# ============================================================

print("\nSkewness")
print(df.skew(numeric_only=True))
```
